Task 5. Generating word cloud from the first 50 words from the text.

```python
from wordcloud import WordCloud
from PIL import Image 
import urllib.request 
import numpy as np 
text2=urllib.request.urlopen("https://raw.githubusercontent.com/annikakalvet/dt-2025-data-visualisation/refs/heads/codespace-jubilant-orbit-gr97595599x39g76/notebooks/Assignment/Week11Homework/text.txt").read().decode("utf-8") #convert the text to a readable format (last one)
words=text2.split(" ")
img=urllib.request.urlretrieve("https://drive.google.com/uc?export=download&id=1bkSZHxaK3rgcJGoAoobMUAs0QjZS2XGM","image1.png")
def color_function(word, font_size, position, orientation, random_state=None, **additional_args):
   if len(word) < 6:       
       return "green"       
   return "purple"            
WordCloud(mask=np.array(Image.open("image1.png")), stopwords=["ja","ei", "aga", "kui", "või", "on"], background_color="white").generate_from_text(text2).recolor(color_func=color_function).to_image()
```

Task 6. This generates word cloud from the text, sees the shape of the image and keeps the word cloud in the same shape. It will color the words under 6 letters long to be green and other words purple. It also removes the most common words that don't add any value to the word cloud. 

Analysis of the word cloud. It appears that the most used words are 'eesti', 'keel(t)', 'pärandkeel*', and 'laps*'. These are mostly green, implicating that shorter words are used more often. As the article was about heritage language in Estonia and how it's obtained by children, all of the most common words make sense. I suppose Estonian is a bit tricky language to make a word cloud from because there are a lot of forms from the same word - the code reads them as different words, while they actually often mark the same meaning. In this word cloud we can see alot of forms from the same word, making the cloud a bit more crowded with information that it maybe should be.
