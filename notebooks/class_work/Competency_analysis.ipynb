```python
import pandas as pd


df=pd.read_csv("Competency_analysis.csv")
df
```

```python
import matplotlib.pyplot as plt


df.plot(kind="bar", x="Student", y="Hours_Studied", legend=False)
plt.ylabel("Hours Studied")
plt.title("Effort per student")
plt.show
```

```python
avg_Student=df.groupby ("Student")["Hours_Studied"]. mean()
avg_Student.plot(kind="bar")
plt.title("Most effort per student")
plt.ylabel("Average hours studied")
plt.show()
```

```python
avg_Student=df.groupby ("Student")["Grade"]. mean(). sort_values(ascending=False)
avg_Student.plot(kind="bar")
plt.title("Performance per student")
plt.ylabel("Grade")
plt.show()
```

```python
avg_Student=df.groupby ("Student")["Grade"]. mean()
plt.figure(figsize=(7, 7))
avg_Student.plot(kind="pie", autopct='%1.1f%%')
plt.title("Performance per student")
plt.show()
```

```python
plt.scatter(df["Hours_Studied"], df["Grade"], s=df["Hours_Studied"]*30, alpha=0.6)
plt.title("Studying and grade correlation")
plt.xlabel("Hours Studied")
plt.ylabel("Grade")
plt.show
```

```python
hours_min=df["Hours_Studied"].min()
hours_ql=df["Hours_Studied"].quantile(0.25)
print("Minimum hours:", hours_min, "Q1:", hours_ql)
plt.figure(figsize=(8,6))
df[["Hours_Studied", "Grade"]].boxplot()
plt.title("Grades and hours studied")
plt.ylabel("Values")
plt.show()
```
