```python
sentence="terehommikustsulle"
#extract forst 3 letters, võib ka [:3]
print("Print the first 3 letters:",sentence[0:3])
#no start from zero, so the 5th letter is actually 4th
print("Print letter 5-9:", sentence[5:9]);
#extract the letters from 5th to the end
print("Print 5th to the end:", sentence[5:]);
#extract the last letter
print("Print the last letter:", sentence[-1]);
#extract the last 4 letters
print("The last 4 letters:", sentence[-4:])
#extract first letter of the substring starting from 5
print("The first letter from the second part:", sentence[5:][0])
#extract the first 5 letters from the substring starting from 5
print("5 letters from the second half:", sentence[5:][:5])
#extract entire text (string)
print(sentence[:])
#extract every second character
print("Every second character:", sentence[::2])
```

```python
sentence2="Tere hommikust sulle sellel ilusal päikselisel päeval"
words=sentence2.split(" ")
print("The words:", *words)
print("The first word:", words[0])
#extarct the first 3 words
print("The first three words:", *words[:3])
#extract last 3 words
print("The last 3 words:", *words[-3:])
print("The no of words:", len(words))
```
